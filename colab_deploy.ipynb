```
import os
import shutil
from IPython.display import clear_output

#@title <h1><B><font color=red>𝗦𝗣𝗦𝗥-𝗩𝟯 𝗕𝗢𝗧</B>
# @markdown <div><center><a href="https://github.com/v88409/spsr-v3"><img height="100" src="https://opengraph.githubassets.com/1/v88409/spsr-v3"></center></a></div>
# @markdown <br><center><h2><strong><font color=red>Save Restricted Content Bot v3</strong></h2></center>

# GitHub repository URL
GITHUB_URL = "https://github.com/v88409/spsr-v3.git"
BRANCH_NAME = "main"

# Base directory for cloning
base_dir = './repo'

def clone_or_update_repo(repo_url, base_directory, branch_name):
    repo_name = os.path.basename(repo_url).replace('.git', '')
    project_dir = os.path.join(base_directory, repo_name)

    if os.path.exists(project_dir):
        print(f"Deleting existing repository at: {project_dir} ...")
        shutil.rmtree(project_dir)
        print("Deleted existing repository successfully!")

    print(f"Cloning repository from {repo_url} (branch: {branch_name})...")
    clone_cmd = f"git clone -b {branch_name} {repo_url} {project_dir}"
    os.system(clone_cmd)
    print("Repository cloned successfully!")

    return project_dir

project_dir = clone_or_update_repo(GITHUB_URL, base_dir, BRANCH_NAME)

try:
    print(f"Entering project directory: {os.path.basename(project_dir)}...")
    os.chdir(project_dir)
    print("Entered project directory successfully!")
except FileNotFoundError as e:
    print(f"Error: {e}")
    raise

clear_output()

# Install required dependencies
PIP_INSTALL = "requirements.txt"
print("Installing required dependencies...")
!pip install -q -r {PIP_INSTALL}
print("All requirements installed successfully!")

clear_output()

# Install FFMPEG
print("Installing FFMPEG...")
os.system("apt-get install ffmpeg -qq")
print("FFMPEG installed successfully!")

#@markdown ### <font color=ORANGE>Environment Variables — Mandatory

API_ID = ""  #@param {type:"string"}
os.environ['API_ID'] = API_ID

API_HASH = ""  #@param {type:"string"}
os.environ['API_HASH'] = API_HASH

BOT_TOKEN = ""  #@param {type:"string"}
os.environ['BOT_TOKEN'] = BOT_TOKEN

MONGO_DB = ""  #@param {type:"string"}
os.environ['MONGO_DB'] = MONGO_DB

OWNER_ID = ""  #@param {type:"string"}
os.environ['OWNER_ID'] = OWNER_ID

#@markdown ### <font color=ORANGE>Environment Variables — Optional (khaali chhod sakte ho)

DB_NAME = "telegram_downloader"  #@param {type:"string"}
os.environ['DB_NAME'] = DB_NAME

LOG_GROUP = "-1001234456"  #@param {type:"string"}
os.environ['LOG_GROUP'] = LOG_GROUP

FORCE_SUB = "-1001234567890"  #@param {type:"string"}
os.environ['FORCE_SUB'] = FORCE_SUB

MASTER_KEY = ""  #@param {type:"string"}
if MASTER_KEY:
    os.environ['MASTER_KEY'] = MASTER_KEY

IV_KEY = ""  #@param {type:"string"}
if IV_KEY:
    os.environ['IV_KEY'] = IV_KEY

ADMIN_CONTACT = "https://t.me/username_of_admin"  #@param {type:"string"}
os.environ['ADMIN_CONTACT'] = ADMIN_CONTACT

JOIN_LINK = "https://t.me/team_spy_pro"  #@param {type:"string"}
os.environ['JOIN_LINK'] = JOIN_LINK

clear_output()

# Command to run the bot (this will block and stream logs live —
# this cell keeps the Colab session alive automatically while it runs.
# To stop the bot, press the ■ stop button on this cell.)
RUN_COMMAND = "python3 main.py"
print(f"Running command: {RUN_COMMAND} ...")
!{RUN_COMMAND}
print("Execution completed!")
```
